```python
from pathlib import Path

# Re-define OUTPUT_DIR as it was not found in the current execution scope.
# This assumes Google Drive is already mounted at "/content/drive".
OUTPUT_DIR = Path("/content/drive/MyDrive/QuranReels/output")

rendered = list(OUTPUT_DIR.glob("video_*.mp4"))
print(f"🎉 Total Rendered Videos on Google Drive: {len(rendered)}")
for v in sorted(rendered)[:15]:
    size_mb = v.stat().st_size / (1024*1024)
    print(f" - {v.name} ({size_mb:.2f} MB)")
```
