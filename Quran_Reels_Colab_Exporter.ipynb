---
### ⚡ Optimizations Built-In:
- 🚀 **Local NVMe SSD Buffer**: Renders on high-speed local disk first, then instantly copies to Google Drive (avoids Google Drive network lag during rendering).
- 🧠 **Pre-computed Segmentation**: Engine loads the Quran dataset once into memory instead of reloading for every single video.
- 🔄 **Auto-Resume & Zero Data Loss**: Automatically detects and skips existing videos on Google Drive.
- 📊 **Live Interactive Progress Bar**: Real-time progress bar (`tqdm`) with ETA and time per video.
